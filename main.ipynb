| Metric | High Predictive Value For... | Primary Weakness |
|--------|------------------------------|------------------|
| P/B | Identifying "floor" value in Banks. | Ignores intellectual property/brands. |
| P/E | Long-term market valuation cycles. | Can stay "irrational" for years. |
| PEG | Identifying undervalued growth. | Relies on estimated future growth. |
| Yield | Total return and downside protection. | High yields can signal a dying business. |
